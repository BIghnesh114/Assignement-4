```python
import pandas as pd
import numpy as np
```

# nominal dataset

```python
data= pd.DataFrame({'color':['red','blue','red','green'],'city':['delhi','mumbai','delhi','kolkata']})
print("NOMINAL DATASET")
print(data)
```

```python
n=len(data)
matrix=np.zeros((n,n))
for i in range(n):
  for j in range(n):
    count=0
    for col in data.columns:
      if data.loc[i,col] !=data.loc[j,col]:
        count += 1
    matrix[i,j]=count/len(data.columns)
```

```python
print("\nDISSIMILARITY MATRIX-NOMINAL")
print(pd.DataFrame(matrix,index=['01','02','03','04'],columns=['01','02','03','04']).round(2))
```

# Binary dataset


```python
data=pd.DataFrame({'Student':[1,1,0,0],'Employed':[0,1,1,0],'Purchased':[1,1,0,1]})
```

```python
print("BINARY DATASET")
print(data)
```

```python
n=len(data)
matrix=np.zeros((n,n))
for i in range(n):
  for j in range(n):
    diff=0
    for col in data.columns:
      if data.loc[i,col] !=data.loc[j,col]:
        diff+=1
    matrix[i,j]=diff/len(data.columns)
```

```python
print("\nDISSIMILARITY MATRIX-BINARY")
print(pd.DataFrame(matrix,index=['01','02','03','04'],columns=['01','02','03','04']).round(2))
```

# Ordinal dataset


```python
data=pd.DataFrame({'satisfaction':['Low','Medium','High','Medium']})
print("ORDINAL DATASET")
print(data)
```

### convert ordinal values into ranks:

```python
rank={
    'Low':0,
    'Medium':1,
    'High':2
}
n=len(data)
matrix=np.zeros((n,n))
for i in range(n):
  for j in range(n):
    x=rank[data.loc[i,'satisfaction']]
    y=rank[data.loc[j,'satisfaction']]
```

# Ordinal dissimilarity

```python
matrix[i,j]=abs(x-y)/2
```

```python
print("\nDISSIMILARITY MATRIX-ORDINAL")
print(pd.DataFrame(matrix,index=['01','02','03','04'],columns=['01','02','03','04']).round(2))
```
